```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
)

from pathlib import Path
import joblib
```

## Baseline Machine Learning Pipeline

In this section, we will:

1. Separate features and target
2. Split the data into training and testing sets
3. Build a preprocessing and classification pipeline
4. Train a Logistic Regression model
5. Evaluate the model
6. Make a sample prediction
7. Save and reload the fitted pipeline

```python
X = df.drop(columns=["target"])
y = df["target"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y,
)

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

```python
print("Training class distribution:")
print(y_train.value_counts(normalize=True).sort_index())

print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True).sort_index())
```

```python
pipeline = Pipeline(
    steps=[
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(max_iter=1000)),
    ]
)

pipeline
```

```python
pipeline.fit(X_train, y_train)

print("Model training completed successfully.")
```

```python
y_pred = pipeline.predict(X_test)

print("Number of predictions:", len(y_pred))
print("First 10 predictions:", y_pred[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
)

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
report = classification_report(
    y_test,
    y_pred,
    target_names=cancer.target_names,
)

print(report)
```

```python
sample = X_test.iloc[[0]]

sample_prediction = pipeline.predict(sample)[0]

predicted_class = cancer.target_names[sample_prediction]

print("Predicted class index:", sample_prediction)
print("Predicted class name :", predicted_class)
```

```python
actual_class_index = y_test.iloc[0]
actual_class_name = cancer.target_names[actual_class_index]

print("Predicted:", predicted_class)
print("Actual   :", actual_class_name)
```

```python
current_path = Path.cwd()

if current_path.name == "notebooks":
    project_root = current_path.parent
else:
    project_root = current_path

model_path = project_root / "models" / "breast_cancer_pipeline.joblib"

print("Project root:", project_root)
print("Model path  :", model_path)
```

```python
model_path.parent.mkdir(parents=True, exist_ok=True)

joblib.dump(pipeline, model_path)

print(f"Pipeline saved successfully to: {model_path}")
```

```python
loaded_pipeline = joblib.load(model_path)

print("Pipeline loaded successfully.")
```

```python
loaded_prediction = loaded_pipeline.predict(sample)[0]

loaded_prediction_name = cancer.target_names[loaded_prediction]

print("Prediction from loaded pipeline:", loaded_prediction_name)
print("Original prediction            :", predicted_class)
```

```python
assert loaded_prediction == sample_prediction

print("Saved and reloaded pipeline produces the same prediction.")
```
